# KNN com o conjunto de dados Iris

Sequência básica para carregar os dados, separar treino e teste, treinar, prever e avaliar um classificador KNN.

## 1. Importar as bibliotecas

```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

## 2. Carregar os dados

```python
df = pd.read_csv("clientes_knn.csv")
X = df.drop("segmento", axis=1)
y = df["segmento"]
```

## 3. Separar os dados de treino e teste

```python
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y,
)
```

## 4. Criar o pipeline

O `StandardScaler` é ajustado somente com os dados de treino. Depois, o KNN utiliza os cinco vizinhos mais próximos.

```python
modelo = make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5),
)
```

## 5. Treinar o modelo

```python
modelo.fit(X_treino, y_treino)
```

## 6. Fazer as previsões

```python
y_previsto = modelo.predict(X_teste)
```

## 7. Avaliar o modelo

```python
acuracia = accuracy_score(y_teste, y_previsto)
matriz_confusao = confusion_matrix(y_teste, y_previsto)
relatorio = classification_report(
    y_teste,
    y_previsto,
    target_names=df["segmento"].unique().tolist(),
    digits=3,
)
```

```python
print(f"Acurácia: {acuracia:.4f}")
print("\nMatriz de confusão:")
print(matriz_confusao)
print("\nRelatório de classificação:")
print(relatorio)
```

## 8. Classificar uma nova flor

Informe as medidas, em centímetros, na ordem: comprimento da sépala, largura da sépala, comprimento da pétala e largura da pétala.

```python
#comprimento da sépala, largura da sépala, comprimento da pétala, largura da pétala
novo_cliente = pd.DataFrame([[40, 7200, 6, 230, 7, 24]], columns=X.columns)
classe_prevista = modelo.predict(novo_cliente)
print("Espécie prevista:", classe_prevista)
```
