```python
import pandas as pd
from sklearn.datasets import load_wine, load_breast_cancer
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

SEMENTE = 42
```

```python
# Carregue o dataset Wine como DataFrame usando load_wine(as_frame=True).
# Guarde o objeto retornado em wine e a tabela completa em df.
wine = load_wine(as_frame=True)
df = wine.frame
```

```python
# Mostre a quantidade de linhas e colunas de df usando o atributo shape.
df.shape
```

```python
# Mostre o tipo de dado de cada coluna usando o atributo dtypes.
df.dtypes
```

```python
# Crie X removendo a coluna target de df.
# Crie y contendo somente a coluna target.
X = df.drop(columns="target")
y = df["target"]
```

```python
# Monte um DataFrame chamado dicionario com três colunas:
# atributo: nomes das colunas de X; tipo: tipos convertidos para texto;
# exemplo: primeiro valor encontrado em cada atributo. Depois, exiba o resultado.
dicionario = pd.DataFrame({
    "atributo": X.columns,
    "tipo": [str(t) for t in X.dtypes],
    "exemplo": [X[c].iloc[0] for c in X.columns],
})
print("\nDicionario de dados:")
print(dicionario.to_string(index=False))
```

```python
# Exiba separadamente as dimensões de X e de y usando shape.
print(f"X.shape = {X.shape}  -> {X.shape[0]} amostras, {X.shape[1]} atributos")
print(f"y.shape = {y.shape}  -> {y.shape[0]} rotulos")
```

```python
# Conte quantos exemplos existem em cada classe de y.
# Ordene o resultado pelo número da classe antes de exibi-lo.
print(y.value_counts().sort_index())
```

```python
# Separe X e y em treino e teste. Use 30% para teste, a constante SEMENTE
# como random_state e stratify=y para conservar a proporção das classes.
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y, test_size=0.30, random_state=SEMENTE, stratify=y,
)
```

```python
# Crie e treine um KNeighborsClassifier com 5 vizinhos, sem padronização.
# Faça previsões no teste, calcule a acurácia e mostre o resultado.
sem_padronizacao = KNeighborsClassifier(n_neighbors=5)
sem_padronizacao.fit(X_treino, y_treino)
ac_sem = accuracy_score(y_teste, sem_padronizacao.predict(X_teste))
```

```python
# Crie um pipeline com StandardScaler e KNeighborsClassifier com 5 vizinhos.
# Treine, faça previsões no teste, calcule a acurácia e mostre o resultado.
com_padronizacao = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
com_padronizacao.fit(X_treino, y_treino)
ac_com = accuracy_score(y_teste, com_padronizacao.predict(X_teste))
```
